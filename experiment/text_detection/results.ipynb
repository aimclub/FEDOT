```python
from pathlib import Path
import pandas as pd
from IPython.display import display

reports = Path('experiment/text_detection/reports')
summary = pd.read_csv(reports / 'benchmark_summary.csv')
pd.set_option('display.max_columns', None)
display(summary.round(4))
```

```python
import matplotlib.pyplot as plt

axis = summary.plot(x='detector', y=['precision', 'recall', 'f1'],
                    kind='bar', rot=0, figsize=(11, 5), ylim=(0, 1.12))
for bars in axis.containers:
    axis.bar_label(bars, fmt='%.3f', padding=3)
axis.set_ylabel('Score')
plt.tight_layout()
plt.show()
```

```python
errors = pd.read_csv(reports / 'benchmark_errors.csv')
display(errors.loc[errors['detector'] == 'improved_v2'])
```

```python
by_dataset = pd.read_csv(reports / 'benchmark_per_dataset.csv')
display(by_dataset.loc[by_dataset['detector'] == 'improved_v2']
        .sort_values('f1')[['dataset', 'columns', 'precision', 'recall', 'f1']])
```
