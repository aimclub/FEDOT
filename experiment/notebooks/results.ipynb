```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
root = next(
    p for p in [Path.cwd(), *Path.cwd().parents]
    if (p / "experiment" / "reports").exists()
)
split = "train"

reports = root / "experiment" / "reports"

if split == "test":
    reports = reports / "test"

summary = pd.read_csv(reports / "benchmark_summary.csv")
errors = pd.read_csv(reports / "benchmark_errors.csv")
per_dataset = pd.read_csv(reports / "benchmark_per_dataset.csv")

display(summary.round(4))
```

```python
table = summary.rename(columns={
    "detector": "Detector",
    "columns": "Total Columns",
    "text_columns": "Text Columns",
    "tp": "TP",
    "fp": "FP",
    "fn": "FN",
    "accuracy": "Accuracy",
    "precision": "Precision",
    "recall": "Recall",
    "f1": "F1 Score"
})

display(
    table[["Detector", "Precision", "Recall", "F1 Score"]]
    .style.format(precision=4)
    .hide(axis="index")
)
```

```python
ax = summary.plot(
    x="detector",
    y=["precision", "recall", "f1"],
    kind="bar",
    figsize=(11, 5),
    rot=0,
    ylim=(0, 1.1),
    title="benchmark Results"
)
for container in ax.containers:
    ax.bar_label(
        container,
        labels=[f"{bar.get_height().round(4)}" for bar in container],
        padding=3,
        fontsize=9
    )

plt.ylabel("Score")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
